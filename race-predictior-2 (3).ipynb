## Functions to parse particular columns 

```python
def fix_temperature(df_temp):
    # Średnia temperatura z max i min
    if {'Max Temperature', 'Min Temperature'}.issubset(df_temp.columns):
        # Konwersja z przecinków na kropki, błędne wartości -> NaN
        for col in ['Max Temperature', 'Min Temperature']:
            df_temp[col] = (
                df_temp[col]
                .astype(str)
                .str.replace(',', '.', regex=False)
                .replace(['', 'nan', 'None'], np.nan)
                .astype(float)
            )        
        # Oblicz średnią tylko gdy coś istnieje
        valid = df_temp[['Max Temperature', 'Min Temperature']].notna().any(axis=1)
        df_temp.loc[valid, 'Average Temperature'] = df_temp[['Max Temperature', 'Min Temperature']].mean(axis=1)
        df_temp.drop(['Max Temperature', 'Min Temperature'], axis=1, inplace=True)
    else:
        print("Missing 'Max Temperature' or 'Min Temperature' columns")
    return df_temp

def parse_avg_speed(row):
    val = row.get("Average Speed")
    typ = row.get("Activity Type")
    
    # Brakujace wartości i '--'
    if val in [None, "", "--"] or pd.isna(val):
        return np.nan
    
    val = str(val).replace(",", ".")
    
    try:
        # Kolarskwo: już w km/h
        if typ == "Ride":
            kmh = float(val)
        
        # Bieganie: min/km → km/h
        elif typ == "Run":
            if ":" in val:
                mins, secs = map(float, val.split(":"))
            else:
                mins = float(val)
                secs = 0
            pace_min_per_km = mins + secs/60
            kmh = 60 / pace_min_per_km
        
        # Pływanie: min/100m → km/h
        elif typ == "Swim":
            if ":" in val:
                mins, secs = map(float, val.split(":"))
            else:
                mins = float(val)
                secs = 0
            pace_min_per_100m = mins + secs/60
            kmh = 60 / pace_min_per_100m * 10
        
        # Inne aktywności -> NaN bo niepotrzebne
        else:
            return np.nan
        
        # Zamień km/h → m/s
        return kmh * 1000 / 3600
    
    except:
        return np.nan


def fix_swim(df_garmin):
    
    df_garmin["Distance"] = pd.to_numeric(df_garmin["Distance"], errors="coerce")
    
    # Jeśli dystans za duży (podany w metrach) zmiana jednostki 
    mask_swim_fix = (df_garmin["Activity Type"] == "Swim") & (df_garmin["Distance"] > 15)
    df_garmin.loc[mask_swim_fix, "Distance"] = df_garmin.loc[mask_swim_fix, "Distance"] / 1000
    
    print(f"Fixed {mask_swim_fix.sum()} swim distances bigger than 15 (converted to km)")
    return df_garmin
```

## Finding max HR for each athlete

```python
def validate_hr_data(df):
    # Usuwamy fizycznie niemożliwe wartości tętna 
    hr_mask = (df['Average Heart Rate'] >= 40) & (df['Average Heart Rate'] <= 220)
    return df[hr_mask]

df_final = validate_hr_data(df_final)
max_hr_per_athlete = df_final.groupby("athlete")["Max Heart Rate"].max()
print(max_hr_per_athlete)
```

```python
# Usuń zawodników, którzy nie mają żadnych danych o tętnie
def remove_athletes_without_hr(df):
    athletes_to_remove = (
        df.groupby("athlete")[["Max Heart Rate", "Average Heart Rate"]]
        .apply(lambda x: x.isna().all().all())
    )
    athletes_to_remove = athletes_to_remove[athletes_to_remove].index.tolist()

    if athletes_to_remove:
        print(f"Usuwam zawodników bez danych o tętnie: {athletes_to_remove}")
        df = df[~df["athlete"].isin(athletes_to_remove)]
    else:
        print("Wszyscy zawodnicy mają dane o tętnie")
    return df


# Użycie po scaleniu wszystkich danych
df_final = remove_athletes_without_hr(df_final)
```

```python
# Filtrujemy aktywności biegowe/rowerowe
df_no_swim = df_final[df_final["Activity Type"].isin(['Ride', 'Run'])]

# Grupujemy po sportowcu i wybieramy 10 najwyższe max_hr
top10_avg_hr = (
    df_no_swim.sort_values(["athlete", "Max Heart Rate"], ascending=[True, False])
    .groupby("athlete")["Max Heart Rate"]
    .apply(lambda x: x.head(10).mean())  # średnia z 10 najwyższych
)

# Konwersja do słownika
max_hr_dict = top10_avg_hr.to_dict()
print(max_hr_dict)
```

## Handle missing values

```python
# Sprawdzenie missing values 
missing = df_final.isnull().sum()
print("Missing values before imputation:\n", missing)

# Kolumny numeryczne: wypełnij mediana dla sportowca
numeric_cols = df_final.select_dtypes(include='number').columns

activity_medians = (
    df_final.groupby("Activity Type")[numeric_cols]
    .median()
    .to_dict(orient="index")
)

def fill_missing_by_group(x):
    activity = x.name[1]  # grupowanie po (athlete, Activity Type)
    x_filled = x.fillna(x.median())  # per-athlete+activity mediana
    # medaina per-activity jeśli nadal NaN
    for col in numeric_cols:
        if x_filled[col].isna().any():
            fallback_val = activity_medians.get(activity, {}).get(col, np.nan)
            x_filled[col] = x_filled[col].fillna(fallback_val)
    return x_filled

df_final[numeric_cols] = (
    df_final.groupby(['athlete', 'Activity Type'], group_keys=False)[numeric_cols]
    .apply(fill_missing_by_group)
)

# Kolumny nienumeryczne
object_cols = df_final.select_dtypes(include='object').columns
df_final[object_cols] = df_final[object_cols].fillna('Unknown')

# Sprawdzenie rezultatow
missing_after = df_final.isnull().sum()
print("Missing values after imputation:\n", missing_after)

df_final.head()
```

## Adding HR Max, Activity ID and is_triathlon columns 

```python
df_final['HRMax'] = df_final["athlete"].map(max_hr_dict)
df_final["Activity ID"] = [i+1 for i in range(len(df_final))]


def identify_triathlon_runs(df, max_time_gap_hours=6):
    df = df.copy()
    df['is_triathlon'] = 0
    
    for athlete in df['athlete'].unique():
        athlete_data = df[df['athlete'] == athlete].sort_values('Activity Date')
        
        # Szukamy sekwencji: Ride → Run w określonym czasie
        for i in range(len(athlete_data) - 1):
            current = athlete_data.iloc[i]
            next_act = athlete_data.iloc[i + 1]
            
            if (current['Activity Type'] == 'Ride' and 
                next_act['Activity Type'] == 'Run'):
                
                time_gap = (next_act['Activity Date'] - current['Activity Date']).total_seconds() / 3600
                
                if time_gap <= max_time_gap_hours:
                    # Oznaczamy bieg jako część triathlonu
                    df.loc[next_act.name, 'is_triathlon'] = 1
    
    return df
    
df_final = identify_triathlon_runs(df_final, max_time_gap_hours=6)

# Sprawdzamy poprawność oznaczeń triathlonów
triathlon_runs = df_final[(df_final['is_triathlon'] == 1)]
print(f"Znaleziono {len(triathlon_runs)} biegów triathlonowych")
print("Rozkład po sportowcach:")
print(triathlon_runs['athlete'].value_counts())
```

## Races identification 

```python
def identify_races(df, hr_threshold=0.86, days_window=365):
    df = df.copy()
    common_race_distances = [5000, 10000, 21100, 42195]
    run_mask = df["Activity Type"] == "Run"
    run_df = df[run_mask].sort_values(["athlete", "Activity Date"])
    
    rolling_list = []
    
    for athlete, subdf in run_df.groupby("athlete"):
        subdf["Activity Date"] = pd.to_datetime(subdf["Activity Date"])
        subdf = subdf.sort_values("Activity Date").set_index("Activity Date")
        subdf["avg_distance_runner"] = subdf["Distance"].rolling(f"{days_window}D", min_periods=1).mean()
        subdf["avg_speed_runner"] = subdf["Average Speed"].rolling(f"{days_window}D", min_periods=1).mean()
        rolling_list.append(subdf.reset_index())
    
    rolling_df = pd.concat(rolling_list, ignore_index=True)
    df = df.merge(rolling_df[["Activity ID", "avg_distance_runner", "avg_speed_runner"]], on="Activity ID", how="left")
    
    # Warunki
    cond_hr = df["Average Heart Rate"] > (df["HRMax"] * hr_threshold)
    cond_distance =  df["Distance"] > 40000
    cond_speed = df["Average Speed"] > 1.4 * df["avg_speed_runner"]
    df["time_ratio"] = df["Elapsed Time"] / df["Moving Time"].replace(0, np.nan)
    cond_ratio = df["time_ratio"] <= 1.1
    
    is_triathlon = df["is_triathlon"]
    distance_tolerance = 0.1  # 10% tolerance
    is_common_distance = False
    for race_dist in common_race_distances:
        lower_bound = race_dist * (1 - distance_tolerance)
        upper_bound = race_dist * (1 + distance_tolerance)
        is_common_distance |= (df["Distance"] >= lower_bound) & (df["Distance"] <= upper_bound)

    # Race score
    df["race_score"] = cond_hr.astype(int) + cond_distance.astype(int) + cond_speed.astype(int) + is_triathlon.astype(int)

    # Dodatkowy waruenk: wyścig ma min. 3km
    cond_min_distance = df["Distance"] >= 3000

    df["race_score"] = 2*cond_hr.astype(int) + 2*cond_distance.astype(int) + 2*cond_speed.astype(int) + is_triathlon.astype(int)  + is_common_distance.astype(int)
    
    run_mask = df["Activity Type"] == "Run"
    
    # Oznaczenie wyścigow
    df["is_race"] = run_mask & cond_min_distance & (df["race_score"] >= 2 ) & cond_ratio
    
    print(f"Detected {df['is_race'].sum()} race activities out of {len(df)} total")
    return df

df = identify_races(df_final)
df = df.drop(['avg_distance_runner', 'avg_speed_runner', 'time_ratio', 'race_score'], axis=1)
df
```

## Build training sequences 

```python
from sklearn.model_selection import train_test_split


def build_sequences_and_statics_split(df, days_window=365, test_split=0.2, min_runs=3):
    """
    Returns:
      - train/test split of (X_seq, X_static_df, y, meta)
      ensuring no data leakage between train/test.
    """
    df = df.sort_values(["athlete", "Activity Date"]).reset_index(drop=True)

    numeric_cols = [
        "Distance", "Elapsed Time", "Moving Time", "Average Speed", "Max Speed",
        "Average Heart Rate", "Max Heart Rate", "Elevation Gain", "Elevation Loss",
        "Average Temperature", "is_triathlon"
    ]
    df[numeric_cols] = df[numeric_cols].astype(np.float32)

    seqs, statics, ys, metas, splits = [], [], [], [], []

    # grupowanie per athlete
    for athlete, sub in df.groupby("athlete"):
        sub = sub.sort_values("Activity Date").reset_index(drop=True)
        runs = sub[sub["Activity Type"].str.lower() == "run"].copy()
        races = runs[runs["is_race"]].copy()
        if races.empty:
            continue

        # Podział na train/test chronologicznie by zapobiec wyciekowi danych testowych 
        races = races.sort_values("Activity Date")
        n_test = max(1, int(len(races) * test_split))
        test_races = races.tail(n_test)
        train_races = races.head(len(races) - n_test)

        for r, split in [*[(r, "train") for _, r in train_races.iterrows()],
                         *[(r, "test")  for _, r in test_races.iterrows()]]:
            race_date = r["Activity Date"]
        
            # Tylko treningi przez wyścigiem (brak wycieku danych)
            mask = (runs["Activity Date"] >= race_date - pd.Timedelta(days=days_window)) & \
                   (runs["Activity Date"] < race_date)
            window_runs = runs.loc[mask].copy()
            if len(window_runs) < min_runs:
                continue

            window_runs["days_to_race"] = (race_date - window_runs["Activity Date"]).dt.days

            seq_numeric_feats = [
                "days_to_race", "Distance", "Elapsed Time", "Moving Time",
                "Average Speed", "Max Speed", "Average Heart Rate", "Max Heart Rate",
                "Elevation Gain", "Elevation Loss", "Average Temperature"
            ]
                    
            # Osobne arrays dla numeric  features
            seq_array = window_runs[seq_numeric_feats].to_numpy(dtype=np.float32)

            # Agregacja statystyk 
            window_others = sub[(sub["Activity Type"].str.lower() != "run") &
                                (sub["Activity Date"] >= race_date - pd.Timedelta(days=days_window)) &
                                (sub["Activity Date"] <= race_date)]

            def agg_activity(df_, names):
                sel = df_["Activity Type"].str.lower().isin(names)
                return (
                    float(df_.loc[sel, "Elapsed Time"].sum(min_count=1) or 0.0),
                    float(df_.loc[sel, "Distance"].sum(min_count=1) or 0.0),
                    int(sel.sum())
                )

            swim_t, swim_dist, swim_sessions = agg_activity(window_others, ["swim"])
            bike_t, bike_dist, bike_sessions = agg_activity(window_others, ["ride"])
            gym_t, _, gym_sessions = agg_activity(window_others, ["weight training"])

            years_of_exp = (
                (runs["Activity Date"].max() - runs["Activity Date"].min()).days / 365.0
                if runs.shape[0] > 1 else 0.0
            )

            static = {
                "athlete": athlete,
                "lead_to_race_date": race_date,
                "years_of_experience": years_of_exp,
                "swim_t_total": swim_t,
                "swim_dist_total": swim_dist,
                "swim_sessions": swim_sessions,
                "bike_t_total": bike_t,
                "bike_dist_total": bike_dist,
                "bike_sessions": bike_sessions,
                "gym_t_total": gym_t,
                "gym_sessions": gym_sessions,
                "race_distance_m": float(r["Distance"]),
                "race_elevation_gain_m": float(r["Elevation Gain"]),
                "is_triathlon": float(r["is_triathlon"]),
                "split": split
            }

            # Przewidywane tempo (sec/km)
            y_val = r.get("Moving Time", r.get("Elapsed Time", None))
            if pd.isna(y_val) or y_val is None:
                continue
            pace_sec_per_km = float(y_val) / (float(r["Distance"]) / 1000.0)

            seqs.append(seq_array)
            statics.append(static)
            ys.append(pace_sec_per_km)
            metas.append({
                "athlete": athlete,
                "lead_to_race_date": race_date,
                "race_activity_id": r.get("Activity ID"),
                "race_distance_m": float(r["Distance"]),
                "split": split
            })
            splits.append(split)

    # Grupowanie na test/train 
    df_stat = pd.DataFrame(statics)
    df_meta = pd.DataFrame(metas)
    y = np.array(ys, dtype=np.float32)

    train_mask = df_stat["split"] == "train"
    test_mask  = df_stat["split"] == "test"

    train_seqs = [s for s, sp in zip(seqs, splits) if sp == "train"]
    test_seqs  = [s for s, sp in zip(seqs, splits) if sp == "test"]

    train_stat_df = df_stat.loc[train_mask].drop(columns="split").reset_index(drop=True)
    test_stat_df  = df_stat.loc[test_mask].drop(columns="split").reset_index(drop=True)

    train_y = y[train_mask.values]
    test_y  = y[test_mask.values]

    train_meta = df_meta.loc[train_mask].reset_index(drop=True)
    test_meta  = df_meta.loc[test_mask].reset_index(drop=True)

    print(f"Built sequences: total={len(seqs)}, train={len(train_seqs)}, test={len(test_seqs)}")
    return (train_seqs, train_stat_df, train_y, train_meta), (test_seqs, test_stat_df, test_y, test_meta)
```

```python
from sklearn.preprocessing import StandardScaler
from tensorflow.keras.preprocessing.sequence import pad_sequences
import joblib

from sklearn.preprocessing import StandardScaler
from tensorflow.keras.preprocessing.sequence import pad_sequences
import joblib

def scale_and_pad(train_seqs, test_seqs, train_stat_df, test_stat_df, train_y, test_y,
                          static_numeric_cols, static_binary_cols,
                          seq_numeric_dim,  
                          pad_value=999.0):
    
    # Dla static features 
    static_scaler = StandardScaler()
    
    # Skaluj tylko numeric features
    X_static_train_numeric = static_scaler.fit_transform(train_stat_df[static_numeric_cols].values)
    X_static_test_numeric = static_scaler.transform(test_stat_df[static_numeric_cols].values)
    
    # Binary features bez skalowania (już są 0/1)
    X_static_train_binary = train_stat_df[static_binary_cols].values.astype(float)
    X_static_test_binary = test_stat_df[static_binary_cols].values.astype(float)
    
    # Połączenie numeric + binary
    X_static_train = np.column_stack([X_static_train_numeric, X_static_train_binary])
    X_static_test = np.column_stack([X_static_test_numeric, X_static_test_binary])

    # Dla sequence features 
    seq_scaler = StandardScaler()
    
    # Połączenie sekwencji do skalowania
    all_sequence_data = np.vstack([s for s in train_seqs if len(s) > 0])
    seq_scaler.fit(all_sequence_data)
    
    train_seqs_scaled = [seq_scaler.transform(s) if len(s) > 0 else np.empty((0, seq_numeric_dim)) 
                        for s in train_seqs]
    test_seqs_scaled = [seq_scaler.transform(s) if len(s) > 0 else np.empty((0, seq_numeric_dim)) 
                       for s in test_seqs]

    # Pad sequences
    max_len = max(max(len(s) for s in train_seqs_scaled), max(len(s) for s in test_seqs_scaled))
    
    X_seq_train = pad_sequences(train_seqs_scaled, maxlen=max_len, padding="post", value=pad_value)
    X_seq_test = pad_sequences(test_seqs_scaled, maxlen=max_len, padding="post", value=pad_value)

    # Scale target
    y_scaler = StandardScaler()
    y_train_scaled = y_scaler.fit_transform(train_y.reshape(-1, 1)).flatten()
    y_test_scaled = y_scaler.transform(test_y.reshape(-1, 1)).flatten()

    scalers = {"seq": seq_scaler, "static": static_scaler, "y": y_scaler}
    
    print(f" Scaled and padded shapes:")
    print(f"   X_seq_train: {X_seq_train.shape}")
    print(f"   X_seq_test:  {X_seq_test.shape}")
    print(f"   X_static_train: {X_static_train.shape} (numeric: {len(static_numeric_cols)}, binary: {len(static_binary_cols)})")
    print(f"   X_static_test:  {X_static_test.shape}")
    print(f"   y_train: {y_train_scaled.shape}")
    print(f"   y_test:  {y_test_scaled.shape}")
    print(f"   Max sequence length: {max_len}")
    
    return X_seq_train, X_seq_test, X_static_train, X_static_test, y_train_scaled, y_test_scaled, scalers
```

# Model 

```python
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, GRU, Dense, Dropout, concatenate, BatchNormalization, Masking, Bidirectional
from tensorflow.keras.optimizers import Adam
from sklearn.metrics import mean_absolute_error, mean_squared_error


def build_gru_race_model(seq_shape, static_shape, hidden_size=16, dense_units=32, dropout_rate=0.3):
    """
    seq_shape: (max_seq_len, n_seq_features)
    static_shape: (n_static_features,)
    """
    # Sequential branch
    seq_input = Input(shape=seq_shape, name="seq_input")
    x = Masking(mask_value=999.0)(seq_input)
    x = Bidirectional(GRU(hidden_size, return_sequences=True, kernel_regularizer="l2", name="gru_1"))(x)
    x = Dropout(dropout_rate)(x)
    x = BatchNormalization()(x)
    x = Bidirectional(GRU(hidden_size, return_sequences=False, kernel_regularizer="l2", name="gru_2"))(x)
    x = Dense(dense_units, activation="relu", kernel_regularizer="l2")(x)
    x = BatchNormalization()(x)
    x = Dropout(dropout_rate)(x)

    # Static branch
    static_input = Input(shape=static_shape, name="static_input")
    s = Dense(dense_units, activation="relu", kernel_regularizer="l2")(static_input)
    s = BatchNormalization()(s)
    s = Dropout(dropout_rate)(s)

    # Merge branches 
    merged = concatenate([x, s])
    merged = Dense(dense_units, activation="relu", kernel_regularizer="l2")(merged)
    merged = BatchNormalization()(merged)
    merged = Dropout(dropout_rate)(merged)

    #Output
    output = Dense(1, activation="linear", name="pace_sec_per_km")(merged)

    # Model
    model = Model(inputs=[seq_input, static_input], outputs=output)
    model.compile(
        optimizer=Adam(learning_rate=8e-4),
        loss="mae",
        metrics=["mae", "mse", "mape"]
    )

    return model
def mean_absolute_percentage_error(y_true, y_pred):
    y_true, y_pred = np.array(y_true), np.array(y_pred)
    # avoid division by zero
    non_zero_mask = y_true != 0
    return np.mean(np.abs((y_true[non_zero_mask] - y_pred[non_zero_mask]) / y_true[non_zero_mask])) * 100
    
def evaluate(y_true, y_pred):
    mae = mean_absolute_error(y_true, y_pred)
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    nrmse = rmse / np.mean(y_true)
    mape = mean_absolute_percentage_error(y_true, y_pred)
    return mae, rmse, nrmse, mape
```

```python
seq_shape = (X_seq_train.shape[1], X_seq_train.shape[2])
static_shape = (X_static_train.shape[1],)

X_static_train = np.nan_to_num(X_static_train, nan=0.0)
X_static_test = np.nan_to_num(X_static_test, nan=0.0)

model = build_gru_race_model(seq_shape, static_shape)
model.summary()
```
